# Task 9: Iterators and Generators
## Create an iterator for the first 5 even numbers.
## Then, create a generator for the same.


```python
# Create an iterator for the first 5 even numbers
class EvenNumberIterator:
    def __init__(self):
        self._data = [0, 2, 4, 6, 8]
        self._index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self._index < len(self._data):
            result = self._data[self._index]
            self._index += 1
            return result
        else:
            raise StopIteration

even_iter = EvenNumberIterator()
for num in even_iter:
    print(num)
```

```python
# create a generator for the same.
def even_number_generator():
    num = 0
    while True:
        yield num
        num += 2
        if num > 8:
            break

even_gen = even_number_generator()
for num in even_gen:
    print(num)
```

# Task 10: Map, Reduce, and Filter
## Use map to square all numbers in a list.
## Use reduce to find the product of all numbers in a list.
## Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

```python
# Use map to square all numbers in a list.

numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
squared_numbers
```

```python
# Use reduce to find the product of all numbers in a list.
from functools import reduce
product = reduce(lambda x, y: x*y, numbers)
product
```

```python
# Use filter to get only even numbers from a list.
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
even_numbers
```

# Task 11: Object-Oriented Programming - Creating a Class
## Create a class called 'Rectangle' with attributes 'length' and 'width'.
## Include methods to calculate area and perimeter.

```python
# Include methods to calculate area and perimeter.
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(5, 3)
print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

# Task 12: Pandas Data Analysis
## Using the following DataFrame, perform these tasks:
## 1. Find the average salary by department
## 2. Get the names of employees with salary > 60000
## 3. Add a new column 'Bonus' which is 10% of salary

```python
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})
df_employees
```

```python
# Find the average salary by department

averageSalary = df_employees.groupby('Department')['Salary'].mean()
averageSalary
```

```python
# Get the names of employees with salary > 60000

employees_high_salary = df_employees[df_employees['Salary'] > 60000]
names_high_salary = employees_high_salary['Name']
print(names_high_salary)
```

```python
# Add a new column 'Bonus' which is 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.1
df_employees
```
